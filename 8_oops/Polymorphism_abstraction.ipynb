Problem: Build a payment-processing example with an abstract Payment class containing a pay() method. Implement CreditCardPayment, UPIPayment, and NetBankingPayment. Demonstrate runtime polymorphism by calling the same method on different payment objects.

To declare abstract class we abstract base class called abc.
ABC stands for Abstract Base Class. It's a class from Python's built-in abc module that lets us define a class which:
Can't be instantiated directly — we can't do Payment(100), only create objects of its subclasses.

LEts take example, we are going to write three classes with same method pay()

CreditCardPayment  → pay() → Credit Card
UPIPayment         → pay() → UPI
NetBankingPayment  → pay() → Net Banking


Python determines at runtime which pay() implementation should execute based on the actual object. Runtime polymorphism occurs when the same method call, such as payment.pay(), produces different behavior depending 


                 Payment (Abstract)
                       |
          +------------+-------------+
          |            |             |
     CreditCard       UPI       NetBanking
      Payment        Payment       Payment
          |            |             |
        pay()        pay()         pay()
        
Payment is abstract (ABC + @abstractmethod) — it can't be instantiated directly, only subclassed.



```python
from abc import ABC, abstractmethod


# Abstract class
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


# Credit Card implementation
class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} using Credit Card")


# UPI implementation
class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} using UPI")


# Net Banking implementation
class NetBankingPayment(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} using Net Banking")


# Create different payment objects
payments = [
    CreditCardPayment(),
    UPIPayment(),
    NetBankingPayment()
]


# Runtime polymorphism
for payment in payments:
    payment.pay(5000)
```
